## Q 5. Find the net revenue generated by the E-Commerce company over all orders placed. (6 marks)

### Ans 5

```python
def Net_Revenue(discounted_price):
  if discounted_price>600:
    Net_Revenue= discounted_price*0.25
    return Net_Revenue
  elif(discounted_price==600 or discounted_price>350):
    Net_Revenue= discounted_price*0.15
    return Net_Revenue
  elif(discounted_price==350 or discounted_price>100):
    Net_Revenue= discounted_price*0.1
    return Net_Revenue
  else:
    Net_Revenue=discounted_price*0.05
    return Net_Revenue
```

```python
ECom_Data['ECom_Net_Revenue'] = ECom_Data['discounted_price'].apply(Net_Revenue)
ECom_Data.head()
```

### The Column 'ECom_Net_Revenue', shows the charges collected for each item based on the charges mentioned above

```python
Total_Net_Revenue = ECom_Data['ECom_Net_Revenue'].sum()
Total_Net_Revenue
```

### The Total_Net_Revenue is the total of Net Revenue for all the orders placed in the given dataset i.e,

## Q 9.Compare performance regionwise

## a)	Draw a lineplot for the monthly net Revenue of E-Commerce Company for each region separately.  (4 marks) 

## b)	Identify the best and the worst performing months for each region. (2 marks)

#### Note: Only those days with actual orders(Order_Date) placed are present in the dataset. Assuming there were no orders on other days

### Ans 9 a)

```python
ECom_Data['Order_Date']  = pd.to_datetime(ECom_Data['Order_Date'])
ECom_Data['month']=ECom_Data['Order_Date'].dt.month
ECom_Data['year'] = ECom_Data['Order_Date'].dt.year

import calendar
ECom_Data['month'] = ECom_Data['month'].apply(lambda x: calendar.month_abbr[x])
```

```python
ECom_Data.head()
```

```python
plt.figure(figsize=(20,7))
sns.lineplot(data=ECom_Data, x='month', y='ECom_Net_Revenue', hue='Region',ci=False);
```

```python
sns.catplot(data=ECom_Data, x="month", y="ECom_Net_Revenue", hue="year", col="Region", kind= 'bar');
```
